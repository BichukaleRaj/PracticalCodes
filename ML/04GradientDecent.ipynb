```python
import time 
import matplotlib.pyplot as plt
```

```python
def function(x): 
 return (x + 3) ** 2
```

```python
def gradient(x): 
    return 2 * (x + 3)
```

```python
x = 2.0 
learning_rate = 0.1 
iterations = 30
```

```python
x_values = [x] 
y_values = [function(x)] 
print("Iteration\t x\t\t y") 
print("-" * 40) 
```

```python
start = time.perf_counter() 
for i in range(iterations): 
    x = x - learning_rate * gradient(x) 
    x_values.append(x) 
    y_values.append(function(x)) 
    print(f"{i + 1}\t\t{x:.6f}\t{function(x):.6f}") 
end = time.perf_counter()
```

```python
print("\nMinimum x:", round(x, 6)) 
print("Minimum y:", round(function(x), 6)) 
print("Execution Time:", round(end - start, 6), "seconds") 
```

```python
x_plot = [i / 100 for i in range(-700, 301)]
y_plot = [function(i) for i in x_plot]

plt.figure(figsize=(8, 5))

plt.plot(x_plot, y_plot, label="y = (x + 3)^2")

plt.plot(
    x_values, y_values, "ro--",
    markersize=4, label="Gradient Descent Path"
)

plt.scatter([-3], [0], color="green", s=100, label="Minimum (-3, 0)")
plt.scatter([2], [25], color="orange", s=100, label="Starting Point (2, 25)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Algorithm")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
